# SQD + SBD walkthrough — H2O

Same self-consistent SQD workflow as `run_sqd_sbd.py`, but interactive and serial.
Runs on a single MPI rank (mpi4py auto-initializes `MPI.COMM_WORLD` with size=1 inside a Jupyter kernel; SBD's MPI collectives become no-ops).

Workload: H2O FCIDUMP (NORB=24, NELEC=10, MS2=0), the 275 bitstrings in `count_dict_h2o.json`, 1 batch, 2 SQD iterations.
Converges to ≈ -76.236 Ha in seconds on CPU — upstream's value for this 75,625-determinant subspace (FCI ≈ -76.2438 Ha). For tighter results use a larger determinant list (`h2o-1em5-alpha.txt` → 5514 bitstrings → 3.0e7 determinants) on a GPU.

**Multi-rank version:** [`run_sqd_sbd.py`](./run_sqd_sbd.py) — same recipe, run with `mpirun -np N python run_sqd_sbd.py …`.

## 1. Imports + environment check

```python
from functools import partial
import json, numpy as np
from pathlib import Path
from mpi4py import MPI
from pyscf import ao2mo, tools
from qiskit_addon_sqd.fermion import diagonalize_fermionic_hamiltonian
from qiskit.primitives import BitArray

import sbd
from sbd.sbd_solver import solve_sci_batch
from sbd.device_config import DeviceConfig, print_device_info

comm = MPI.COMM_WORLD
print(f'SBD version            : {sbd.__version__}')
print(f'Compiled SBD backends  : {sbd.available_backends()}')
print(f'MPI library            : {MPI.Get_library_version().splitlines()[0]}')
print(f'MPI world size         : {comm.Get_size()}    (Jupyter kernel = single process)')
print(f'My rank                : {comm.Get_rank()}')
print()
print_device_info()
```

## 2. Load the Hamiltonian (H2O FCIDUMP)

FCIDUMP carries the one- and two-electron integrals plus orbital count and electron count. PySCF reads it; we extract `hcore` and `eri` for the SQD outer loop.

```python
fcidump_path = '../../vendor/sbd-upstream/data/h2o/fcidump.txt'
norb, nelec_total, ms2 = 24, 10, 0          # h2o constants
num_elec_a = (nelec_total + ms2) // 2        # 5 alpha
num_elec_b = (nelec_total - ms2) // 2        # 5 beta

mf    = tools.fcidump.to_scf(fcidump_path)
hcore = mf.get_hcore()
eri   = ao2mo.restore(1, mf._eri, norb)
nuc   = mf.mol.energy_nuc()
print(f'NORB={norb}, nelec=({num_elec_a},{num_elec_b}), nuc={nuc:.6f}')
```

## 3. Bitstrings — a curated H2O determinant set

`count_dict_h2o.json` holds 275 bitstrings in qiskit-addon-sqd's `[beta | alpha]` layout
(24 bits per half, 5 electrons each), built from the vendored `h2o-1em3-alpha.txt` by reusing
every alpha half-determinant for both spin sectors. SBD then forms the 275 × 275 =
75,625-determinant tensor product — the subspace `run_sbd_diag.py` builds from `--adetfile`.

All of them already carry the target Hamming weight, so none are postselected away; of
uniform-random 48-bit strings only `C(24,5)² / 4²⁴ ≈ 6e-6` would qualify. Counts are all 1 —
a determinant list, not a shot histogram.

*(For hardware output, pass your own file: `run_sqd_sbd.py --counts …`.)*

```python
rng = np.random.default_rng(42)

counts = json.loads(Path('count_dict_h2o.json').read_text())
bit_array = BitArray.from_counts(counts, num_bits=norb*2)

weights = {(k[norb:].count('1'), k[:norb].count('1')) for k in counts}
print(f'Loaded {len(counts)} bitstrings ({norb*2} bits / shot) from count_dict_h2o.json')
print(f'(alpha, beta) Hamming weights present: {weights}  -> target ({num_elec_a}, {num_elec_b})')
print(f'Tensor-product subspace: {len(counts)} x {len(counts)} = {len(counts)**2:_} determinants')
```

## 4. Wire SBD into qiskit-addon-sqd's `sci_solver=` slot

No `sbd.init()` and no `mpi_comm=` needed: `solve_sci_batch` auto-initializes the SBD backend on first call and falls back to `MPI.COMM_WORLD` when `mpi_comm` is omitted (in a Jupyter kernel that's the size-1 communicator).

```python
sbd_config = {
    'method': 0,           # 0 = Davidson
    'eps': 1e-8,           # Davidson convergence tolerance
    'max_it': 10,         # max Davidson iterations per diagonalization
    'max_nb': 10,          # block size
    'do_rdm': 0,           # 0 = density only (sufficient for SQD)
    'do_shuffle': 0,
    'carryover_type': 1,   # singles-only carryover (stable default)
    'ratio': 0.1,
    'threshold': 1e-4,
    'bit_length': 20,   # bits per size_t; RIKEN default (64 is UB in bitadvance)
    # serial: 1×1×1 MPI sub-communicator grid
    'adet_comm_size': 1, 'bdet_comm_size': 1, 'task_comm_size': 1,
}

sbd_solver = partial(
    solve_sci_batch,
    sbd_config=sbd_config,
    device_config=DeviceConfig.cpu(),       # macOS / no-GPU notebook
    fcidump_path=fcidump_path,
)
```

## 5. Run the SQD self-consistent loop

Sample → configuration recovery → subsample → SBD diagonalize → carryover → repeat. 1 batch per iteration, 2 SQD iterations.

```python
history = []
def callback(results):
    history.append(results)
    iteration = len(history)
    for i, r in enumerate(results):
        total_e = r.energy + nuc
        dim = np.prod(r.sci_state.amplitudes.shape)
        # carryover dets selected for the next iteration, not the 75,625 diagonalized
        print(f'  iter {iteration}, batch {i}:  E = {total_e:.10f} Ha   carryover dim = {dim:_}')

result = diagonalize_fermionic_hamiltonian(
    hcore, eri, bit_array,
    norb=norb,
    nelec=(num_elec_a, num_elec_b),
    samples_per_batch=300,
    num_batches=1,
    max_iterations=2,
    sci_solver=sbd_solver,
    symmetrize_spin=True,
    callback=callback,
    seed=rng,
)

print()
print(f'Final SQD energy:  {result.energy + nuc:.10f} Ha   (electronic {result.energy:.10f} + nuc {nuc:.6f})')
print(f'FCI reference   :  ≈ −76.24 Ha')
```

## 6. What's in the result?

`result` is an `SCIResult` dataclass with three fields the chemistry side cares about: `energy` (electronic), `sci_state` (the wavefunction in its CI basis), and `orbital_occupancies` (one-body density on the spin orbitals).

**`sci_state.amplitudes`** is a 2D array `C[i_α, i_β]` of shape `(M, N)` — `M` α-determinants × `N` β-determinants. The full wavefunction is

$|\psi\rangle = \sum_{i,j}\, C_{ij}\, |\,\mathrm{ci\_strs\_a}[i]\rangle \otimes |\,\mathrm{ci\_strs\_b}[j]\rangle$

so the linear (1-D) subspace dimension is `M × N` (this is what we printed as *subspace dim* in the callback).

**`orbital_occupancies`** is a `(α_occ, β_occ)` tuple of two 1-D arrays, each of length `norb`. `α_occ[k] = ⟨ψ|n̂_{k,↑}|ψ⟩`, similarly for β. They feed back into qiskit-addon-sqd's `recover_configurations` between iterations.

```python
psi = result.sci_state
alpha_occ, beta_occ = result.orbital_occupancies

print(f'sci_state.amplitudes.shape : {psi.amplitudes.shape}      # (M, N)  =  (n_α-dets, n_β-dets)')
print(f'sci_state.ci_strs_a.shape  : {psi.ci_strs_a.shape}')
print(f'sci_state.ci_strs_b.shape  : {psi.ci_strs_b.shape}')
print(f'subspace dim (M × N)        : {psi.amplitudes.size}')
print(f'norm of ψ                   : {np.linalg.norm(psi.amplitudes):.10f}')
print()

print(f'orbital_occupancies α shape : {alpha_occ.shape}     # (norb,)')
print(f'orbital_occupancies β shape : {beta_occ.shape}')
print(f'sum α_occ  (should be {num_elec_a}) : {alpha_occ.sum():.6f}')
print(f'sum β_occ  (should be {num_elec_b}) : {beta_occ.sum():.6f}')
print()
print('α-occupancy per spatial orbital:')
print(np.array2string(alpha_occ, precision=4, suppress_small=True))
print('β-occupancy per spatial orbital:')
print(np.array2string(beta_occ, precision=4, suppress_small=True))
```

## Notes

- **Why this works in a notebook**: SBD is MPI-native at the C++ level, but a Jupyter kernel is a single Python process. `mpi4py` auto-initializes MPI with `MPI.COMM_WORLD` of size 1; SBD's collectives all become no-ops; the full SQD loop runs on rank 0.
- **Why a curated counts file**: qiskit-addon-sqd postselects on (5α, 5β), which uniform-random strings almost never satisfy. `count_dict_h2o.json` supplies determinants that do. From iteration 2 on, configuration recovery refines them from the previous diagonalization's occupancies — no `initial_occupancies` needed.
- **Multi-rank**: launch [`run_sqd_sbd.py`](./run_sqd_sbd.py) with `mpirun -np N python …` for production scale.
- **GPU**: pass `DeviceConfig.gpu()` (Thrust) or `DeviceConfig.gpu_omp()` (LLVM offload) on a CUDA-capable machine. Not available on macOS — keep `DeviceConfig.cpu()` here.
